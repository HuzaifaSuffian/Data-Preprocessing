# Trimming

```python
upper_limit = df['cgpa'].quantile(q=0.99)
lower_limit = df['cgpa'].quantile(q=0.01)
```

```python
# Treating these values as outliers
df[(df['cgpa'] > upper_limit) | (df['cgpa'] < lower_limit)]
```

```python
new_df = df[(df['cgpa'] < upper_limit) & (df['cgpa'] > lower_limit)]
```

```python
new_df
```

```python
new_df['cgpa'].plot(kind='box')
```

# Capping or Winsorization

```python
new_df = df['cgpa'].clip(lower_limit,upper_limit)
```

```python
new_df.plot(kind='box')
```
